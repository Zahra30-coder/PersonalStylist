```python
query = "SELECT * FROM Products"

df = pd.read_sql(query, conn)
```

```python
print(df.columns)
```

```python
def create_text(row):
    return f"""
Product Name: {row['prod_name']}

Product Type: {row['product_type_name']}

Product Group: {row['product_group_name']}

Color: {row['colour_group_name']}

Department: {row['department_name']}

Index: {row['index_name']}

Index Group: {row['index_group_name']}

Section: {row['section_name']}

Garment Group: {row['garment_group_name']}

Appearance: {row['graphical_appearance_name']}

Description:
{row['detail_desc']}
"""
```

```python
df["text"] = df.apply(create_text, axis=1)
```

```python
print(df.loc[0, "text"])
```

```python
df["text_length"] = df["text"].str.len()

print(df["text_length"].describe())
```

```python
df[["article_id", "text"]].to_csv("product_documents.csv", index=False)
```
