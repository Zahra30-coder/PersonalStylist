```python
# -----------------------------------
# LangGraph State
# -----------------------------------
class StylistState(TypedDict):

    # form fields
    skin_tone: str
    body_shape: str
    height: str
    preferred_colors: List[str]
    occasion: str
    style_preference: str

    # conversational memory
    messages: List[str]

    # output
    recommendation: str
```

```python
# -----------------------------------
# Gemini
# -----------------------------------

model = genai.GenerativeModel(
    "models/gemini-2.5-flash"
)
```

```python
state = {
    "skin_tone": "wheatish",
    "body_shape": "rectangle",
    "height": "5'4",
    "preferred_colors": ["peach", "maroon"],
    "occasion": "corporate meeting",
    "style_preference": "ankle-length dresses",
    "messages": [],
    "recommendation": ""
}
```

```python
# -----------------------------------
# Extract User Attributes
# -----------------------------------

@observe(name="llm_call")
def stylist_agent(state: StylistState):
    history = "\n".join(state["messages"])
    prompt = f"""
    You are an AI personal stylist.

    User Profile:

    Skin Tone:
    {state['skin_tone']}

    Body Shape:
    {state['body_shape']}

    Height:
    {state['height']}

    Preferred Colors:
    {state['preferred_colors']}

    Occasion:
    {state['occasion']}

    Style Preference:
    {state['style_preference']}

    Conversation History:
    {history}

    Latest User Query:
    {user_query}

    Task:
    Give personalized styling recommendations.

    Include:
    - dress suggestions
    - flattering silhouettes
    - color combinations
    - footwear
    - layering ideas
    - fabric suggestions

    Be elegant and conversational.
    """

    response = model.generate_content(prompt)

    state["recommendation"] = response.text

    return state
```

```python
# -----------------------------------
# Graph
# -----------------------------------

graph = StateGraph(StylistState)

graph.add_node(
    "stylist_agent",
    stylist_agent
)

graph.set_entry_point("stylist_agent")

graph.add_edge(
    "stylist_agent",
    END
)

app = graph.compile()
```

```python
# Get API key
api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("GEMINI_API_KEY not found in .env")

# Configure Gemini
genai.configure(
    api_key=api_key
)

# Create model
model = genai.GenerativeModel(
    "models/gemini-2.5-flash"
)

print("Gemini ready!")
```

## Adding Conversation Memory

```python
class ConversationMemory:

    def __init__(self, window_size=6):

        self.window_size = window_size
        self.recent_messages = []
        self.summary = []

    @observe(name="memory_add")
    def add(self, role, content):
        self.recent_messages.append({
            "role": role,
            "content": content
        })

        while len(self.recent_messages) > self.window_size:
            self.recent_messages.pop(0)
            
    @observe(name="memory_get_history")
    def get_recent_history(self):
        return "\n".join(
            f"{m['role']}: {m['content']}"
            for m in self.recent_messages
        )

    def get_summary(self):
        return self.summary
```

```python
memory = ConversationMemory()
```

```python
WINDOW_SIZE = 6
```

```python

def chat_with_memory(state, user_input):

    # Store user message
    state["messages"].append(
        f"user: {user_input}"
    )

    # Keep only recent messages
    state["messages"] = (
        state["messages"][-WINDOW_SIZE:]
    )

    # Invoke graph
    result = app.invoke(state)

    assistant_reply = result["recommendation"]

    # Store assistant reply
    state["messages"].append(
        f"assistant: {assistant_reply}"
    )

    # Again trim memory window
    state["messages"] = (
        state["messages"][-WINDOW_SIZE:]
    )

    return assistant_reply
```

```python
# -----------------------------------
# Trace LLM Calls
# -----------------------------------
@observe(name="ask_llm")
def ask_llm(state, user_query=""):

    history = memory.get_recent_history()

    prompt = f"""
    You are an AI personal stylist.

    User Profile:
    Skin Tone: {state['skin_tone']}
    Body Shape: {state['body_shape']}
    Height: {state['height']}
    Preferred Colors: {", ".join(state['preferred_colors'])}
    Occasion: {state['occasion']}
    Style Preference: {state['style_preference']}

    Conversation Summary:
    {memory.get_summary()}

    Recent Conversation:
    {history}

    Latest User Request:
    {user_query}

    Instructions:
    - Give personalized styling recommendations.
    - Suggest flattering silhouettes.
    - Recommend suitable colors.
    - Suggest footwear/accessories if relevant.
    """

    try:

        response = model.generate_content(prompt)

        output = response.text


        # Send details to Langfuse
        langfuse_context.update_current_observation(
            input=prompt,
            output=output,
            metadata={
                "model": "gemini-pro",
                "provider": "google"
            }
        )


        return output


    except ResourceExhausted:

        error_message = "Rate limit exceeded. Try again shortly."

        langfuse_context.update_current_observation(
            output=error_message
        )

        return error_message
```
